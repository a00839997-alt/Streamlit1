```python
import pandas as pd 
import matplotlib as plt 
import seaborn as sns 
```

## Phase 1

```python
df_policy = pd.read_csv("/Users/luisjativa/Documents/Data Visualization/COV_VAC_POLICY_2024.csv")
df_uptake = pd.read_csv("/Users/luisjativa/Documents/Data Visualization/COV_VAC_UPTAKE_2024.csv")
df_hosp = pd.read_csv("/Users/luisjativa/Documents/Data Visualization/WHO-COVID-19-global-hosp-icu-data.csv")
df_deaths = pd.read_csv("/Users/luisjativa/Documents/Data Visualization/WHO-COVID-19-global-monthly-death-by-age-data.csv")
```

```python
df_policy.rename(columns={'COUNTRY': 'iso_code'}, inplace=True)
df_uptake.rename(columns={'COUNTRY': 'iso_code'}, inplace=True)
df_hosp.rename(columns={'Country_code': 'iso_code'}, inplace=True)
df_deaths.rename(columns={'Country_code': 'iso_code'}, inplace=True)
```

```python
df_policy['DATE'] = pd.to_datetime(df_policy['DATE'])
df_uptake['DATE'] = pd.to_datetime(df_uptake['DATE'])
df_hosp['Date_reported'] = pd.to_datetime(df_hosp['Date_reported'])
df_deaths['Date'] = pd.to_datetime(df_deaths['Year'].astype(str) + '-' + df_deaths['Month'].astype(str).str.zfill(2) + '-01')
```

```python
df_policy.dropna(subset=['VALUE'], inplace=True)
df_hosp.dropna(subset=['Covid_new_hospitalizations_last_7days', 'Covid_new_icu_admissions_last_7days'], how='all', inplace=True)
df_deaths.dropna(subset=['Deaths'], inplace=True)

print(f"Policy: {len(df_policy)}")
print(f"Uptake: {len(df_uptake)}")
print(f"Hosp: {len(df_hosp)}")
print(f"Deaths: {len(df_deaths)}")
```

## Phase 2

```python
print("Groups in Policy:", df_policy['CATEGORY'].unique()[:5]) # Showing the first 5
print("Groups in Uptake:", df_uptake['GROUP'].unique())
print("Age ranges in Deaths:", df_deaths['Agegroup'].unique())
print("-" * 30)

df_policy_seniors = df_policy[df_policy['CATEGORY'] == 'Older adults'].copy()
df_deaths_seniors = df_deaths[df_deaths['Agegroup'] == '65+'].copy()

df_uptake_seniors = df_uptake[df_uptake['GROUP'] == 'older_adults'].copy()
df_policy_seniors['YearQuarter'] = df_policy_seniors['DATE'].dt.to_period('Q')
df_uptake_seniors['YearQuarter'] = df_uptake_seniors['DATE'].dt.to_period('Q')
df_hosp['YearQuarter'] = df_hosp['Date_reported'].dt.to_period('Q')
df_deaths_seniors['YearQuarter'] = df_deaths_seniors['Date'].dt.to_period('Q')
```

## Export CSV 

```python
df_policy_seniors.to_csv('Policy_Seniors_Clean.csv', index=False)
df_uptake_seniors.to_csv('Uptake_Seniors_Clean.csv', index=False)
df_hosp.to_csv('Hospitalizations_Clean.csv', index=False)
df_deaths_seniors.to_csv('Deaths_Seniors_Clean.csv', index=False)
```
